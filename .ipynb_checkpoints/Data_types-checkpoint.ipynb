### Python support 3 categories of data types 
#### 3. user-defind Class 
##### 1. *Basic Types (Integer , Float , Complex , Boolean  and string).*
##### 2. *container Types (List , Tuples , Sets and Dictionary).*
##### 3. *user-defind Class.* 

- *integer*

```python
print(4)
```

```python
print('le3O9')
```

- *Float*

```python
print(4.5)
```

```python
print(1.7e308)
```

- *Boolean*

```python
print(True)
print(False)
```

- *Complex*

```python
print(4+5j)
```

- *String*

```python
print('Bhopal')
print("Bhopal")
```

- *List*

```python
print([1,2,3,3,4,4,4,5,5,6,6,67,7])
```

- *Tuple*

```python
print(1,2,3,4,5,56,6,)
```

```python
print((1,2,3,4,5,56,6,))
```

- *Sets*

```python
print({1,23,33,99,378,9})
```

- *Dict*

```python
print({"Nitesh", 30, "male"})
```

```python
print({'Name':"Nitesh", "Age":30 , "Gender":'male'})
```

```python
print({
    "name":"Anuraj Rajak",
    "Age":21,
    "Gender": "Male",
    "Contract": 9669539437,
    "City": "Bhopal",
    "States": "MP"
})
```
